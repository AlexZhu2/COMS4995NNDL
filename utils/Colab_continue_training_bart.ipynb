```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from tqdm import tqdm
import matplotlib.pyplot as plt
import numpy as np
from torchvision import transforms

from annotations.COCO_caption_dataset_bart import COCOCaptionDataset
from model.caption_models.bart_CNN_transformer import CNNBARTCaptioningModel
```

```python
print(torch.__version__)
print(np.__version__)
```

```python
# ---------------- CONFIG ----------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
EPOCHS = 20
BATCH_SIZE = 32
EMBED_DIM = 512
MAX_LEN = 35
VAL_SAMPLE_IDX = [0,10,246,93,59]

CHECKPOINT_DIR = "/content/drive/MyDrive/Checkpoints-BART/"
VISUAL_DIR     = "/content/drive/MyDrive/Visualizations-BART/"
STATS_DIR      = "/content/drive/MyDrive/Stats-BART/"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(VISUAL_DIR, exist_ok=True)
os.makedirs(STATS_DIR, exist_ok=True)
```

```python
# ------------------------- Transforms -------------------------
val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225])
])
train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.RandomAffine(0, translate=(0.1, 0.1), scale=(0.5, 1.5), shear=10),
    transforms.RandomCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225])
])
```

```python
# ---------------- DATASET ----------------
train_dataset = COCOCaptionDataset('COMS4995NNDL/annotations/captions_train2017.json', 'train2017', max_length=MAX_LEN, transform=train_transform)
val_dataset   = COCOCaptionDataset('COMS4995NNDL/annotations/captions_val2017.json',   'val2017',   max_length=MAX_LEN, transform=val_transform)

train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader   = torch.utils.data.DataLoader(val_dataset,   batch_size=BATCH_SIZE, shuffle=False)

VOCAB_SIZE = len(train_dataset.tokenizer)
PAD_IDX = train_dataset.tokenizer.pad_token_id
```

```python
# ---------------- MODEL ----------------
model = CNNBARTCaptioningModel(pretrained_model_name="facebook/bart-base", embed_dim=EMBED_DIM).to(DEVICE)
optimizer = optim.Adam(model.parameters(), lr=1e-4)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=3, gamma=0.5)
criterion = nn.CrossEntropyLoss(ignore_index=PAD_IDX)

train_losses, val_losses = [], []
best_val_loss = float("inf")
start_epoch = 0
```

```python
# ---------------- LOAD CHECKPOINT ----------------
checkpoint_path = os.path.join(CHECKPOINT_DIR, "best_model_cnn_bart_captioning.pth")
if os.path.exists(checkpoint_path):
    checkpoint = torch.load(checkpoint_path, map_location=DEVICE)
    model.load_state_dict(checkpoint["model_state_dict"])
    optimizer.load_state_dict(checkpoint["optimizer_state_dict"])
    scheduler.load_state_dict(checkpoint["scheduler_state_dict"])
    train_losses = checkpoint["train_losses"]
    val_losses = checkpoint["val_losses"]
    best_val_loss = checkpoint["best_val_loss"]
    start_epoch = checkpoint["epoch"] + 1
    print(f"Loaded checkpoint from epoch {start_epoch}")
```

```python
from nltk.translate.bleu_score import sentence_bleu, corpus_bleu, SmoothingFunction
smoothie = SmoothingFunction().method4
# ---------------- TRAINING ----------------
for epoch in range(10, EPOCHS): # Needs to change 10 back to start_epoch once the model's details are saved
    model.train()
    total_train_loss = 0.0
    train_pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS} [Train]")

    for batch in train_pbar:
        images    = batch['image'].to(DEVICE)
        input_ids = batch['input_ids'].to(DEVICE)

        tgt_input  = input_ids[:, :-1]
        tgt_output = input_ids[:, 1:]
        decoder_attention_mask = (tgt_input != PAD_IDX)

        logits = model(images, tgt_input, decoder_attention_mask=decoder_attention_mask)
        loss   = criterion(logits.reshape(-1, VOCAB_SIZE), tgt_output.reshape(-1))

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_train_loss += loss.item()
        train_pbar.set_postfix(loss=loss.item())

    train_losses.append(total_train_loss / len(train_loader))


    # 👇 collect ground-truth & predictions for BLEU
    all_references = []
    all_candidates = []

    model.eval()
    total_val_loss = 0.0

    with torch.no_grad():
        val_pbar = tqdm(val_loader, desc=f"Epoch {epoch+1}/{EPOCHS} [Val]")

        for i, batch in enumerate(val_pbar):
            images    = batch['image'].to(DEVICE)
            input_ids = batch['input_ids'].to(DEVICE)

            # 1️⃣ (optional) validation loss using teacher forcing
            tgt_input  = input_ids[:, :-1]
            tgt_output = input_ids[:, 1:]

            logits = model(images, tgt_input)
            loss   = criterion(logits.reshape(-1, VOCAB_SIZE), tgt_output.reshape(-1))
            total_val_loss += loss.item()
            val_pbar.set_postfix(loss=loss.item())

            # 2️⃣ generate captions using model.generate()
            generated_ids = model.generate(images, max_length=35, num_beams=4)  # adjust beam/length if needed

            for j in range(images.size(0)):
                # decode predictions
                pred_text = val_dataset.decode(generated_ids[j])
                gt_text = batch['raw_caption'][j] if isinstance(batch['raw_caption'], list) else batch['raw_caption']

                all_references.append([gt_text.lower().split()])  # ref needs list of tokens
                all_candidates.append(pred_text.lower().split())  # hyp needs tokens

                # visualize selected sample
                global_idx_start = i * val_loader.batch_size
                global_idx = global_idx_start + j
                if global_idx in VAL_SAMPLE_IDX:
                    raw_img = batch['image'][j].cpu()
                    img_np = raw_img.permute(1, 2, 0).numpy()
                    std = np.array([0.229, 0.224, 0.225])
                    mean = np.array([0.485, 0.456, 0.406])
                    img_np = np.clip(img_np * std + mean, 0, 1)

                    plt.figure(figsize=(6,6))
                    plt.imshow(img_np)
                    plt.axis("off")
                    plt.title(f"GT: {gt_text}\nPred: {pred_text}", fontsize=10)
                    plt.savefig(f"/content/drive/MyDrive/Visualizations-BART/epoch_{epoch+1}_sample_{global_idx}.png")
                    plt.close()

                    with open(f"/content/drive/MyDrive/Visualizations-BART/epoch_{epoch+1}.txt", "a") as f:
                        f.write(f"Validation Sample {global_idx}:\nGT: {gt_text}\nPred: {pred_text}\n\n")

    # compute BLEU after all validation batches
    corpus_bleu_score = corpus_bleu(all_references, all_candidates, smoothing_function=smoothie)

    avg_val_loss = total_val_loss / len(val_loader)
    val_losses.append(avg_val_loss)

    print(f"Epoch {epoch+1}: Train Loss = {train_losses[-1]:.4f}, Val Loss = {avg_val_loss:.4f}, BLEU = {corpus_bleu_score:.4f}")

    scheduler.step()

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save({
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "epoch": epoch,
            "train_losses": train_losses,
            "val_losses": val_losses
        }, "/content/drive/MyDrive/Checkpoints-BART/best_model_cnn_bart_captioning.pth")
        print("✅ Saved best model.")

# -------------- Save loss curve after each epoch --------------
    plt.figure(figsize=(8,6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses,   label='Val Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig(f"/content/drive/MyDrive/Stats-BART/loss_curve_epoch_{epoch+1}.png")
    plt.close()

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(
            model.state_dict(),
            "/content/drive/MyDrive/Checkpoints-BART/best_model_cnn_bart_captioning.pth"
        )
        print("✅ Saved best model.")
```

```python
import os
import torch
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

from model.caption_models.bart_CNN_transformer import CNNBARTCaptioningModel  # adjust import path
from transformers import BartTokenizer
from torchvision import transforms
```

```python
# ---------------- CONFIG ----------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
CHECKPOINT_PATH = "/content/drive/MyDrive/Checkpoints-BART/best_model_cnn_bart_captioning.pth"
VISUAL_SAVE_DIR = "/content/sample_data/test_rst"
os.makedirs(VISUAL_SAVE_DIR, exist_ok=True)

CUSTOM_IMAGE_PATHS ="/content/sample_data/test_imgs"
MAX_LEN = 35
```

```python
# ---------------- TRANSFORMS ----------------
image_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225])
])
```

```python
# ---------------- MODEL ----------------
model = CNNBARTCaptioningModel(
    pretrained_model_name="facebook/bart-base",
    embed_dim=512,
    freeze_encoder=True
).to(DEVICE)

checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

tokenizer = BartTokenizer.from_pretrained("facebook/bart-base")

print(f"Loaded model from {CHECKPOINT_PATH}")
```

```python
# ---------------- INFERENCE ----------------
ALL_IMAGE_PATHS = os.listdir(CUSTOM_IMAGE_PATHS)
for img_path in ALL_IMAGE_PATHS:
    try:
        img_pil = Image.open(os.path.join(CUSTOM_IMAGE_PATHS, img_path)).convert("RGB")
    except Exception as e:
        print(f"Failed to load {img_path}: {e}")
        continue

    img_tensor = image_transform(img_pil).unsqueeze(0).to(DEVICE)  # (1, C, H, W)

    with torch.no_grad():
        generated_ids = model.generate(
            img_tensor, max_length=MAX_LEN, num_beams=4
        )
        pred_text = tokenizer.decode(
            generated_ids[0],
            skip_special_tokens=True,
            clean_up_tokenization_spaces=True
        )

    # visualize
    img_np = np.array(img_pil) / 255.0
    img_name = img_path.split(".")[0]

    plt.figure(figsize=(6, 6))
    plt.imshow(img_np)
    plt.axis("off")
    plt.title(f"Pred: {pred_text}", fontsize=12)
    plt.savefig(os.path.join(VISUAL_SAVE_DIR, f"{img_name}_pred.png"))
    plt.close()

    # log text
    with open(os.path.join(VISUAL_SAVE_DIR, "custom_predictions.txt"), "a") as f_out:
        f_out.write(f"{img_name}: Pred: {pred_text}\n\n")

    print(f"Processed {img_path}")

print(f"\n Done! Visualizations saved in {VISUAL_SAVE_DIR}")
```

```python
from nltk.translate.bleu_score import sentence_bleu, corpus_bleu, SmoothingFunction
smoothie = SmoothingFunction().method4

model.eval()
all_references = []
all_candidates = []
total_val_loss = 0.0

with torch.no_grad():
    for batch in val_loader:
        images = batch['image'].to(DEVICE)
        input_ids = batch['input_ids'].to(DEVICE)
        generated_ids = model.generate(images, max_length=MAX_LEN, num_beams=4)
        for j in range(images.size(0)):
            pred_text = val_dataset.decode(generated_ids[j])
            gt_text = batch['raw_caption'][j] if isinstance(batch['raw_caption'], list) else batch['raw_caption']
            all_references.append([gt_text.lower().split()])
            all_candidates.append(pred_text.lower().split())

corpus_bleu_score = corpus_bleu(all_references, all_candidates, smoothing_function=smoothie)
print(f"BLEU Score: {corpus_bleu_score:.4f}")
```
